# Notes

Variables:

- Condition: Rank participant saw themselves in on the leaderboard (1-6)

- LB_Good: “I seem to perform very well compared to the others in the ranking list.” (1 (disagree strongly) - 5 (agree strongly))

- LB_Poor: “I seem to perform very poorly compared to the others in the ranking list.” (1 (disagree strongly) - 5 (agree strongly))

- LB_Motivated: “I feel motivated to improve my physical activity (steps or minutes of exercise).” (1 (disagree strongly) - 5 (agree strongly))

- Age: Age of participant

- Gender: Gender of participant

# Setup

```r
#| output: false

# Libraries
library(tidyverse)
library(gtsummary)
library(gt)
library(apaTables)
library(IRdisplay)
IRdisplay::display_html('<style>.container { width:100% !important; }</style>')
```

```r
#| echo: false

# Load data
d = read.csv("/Users/rajharsora/Downloads/csv files/DHS cleaned.csv") %>%
# Include only needed columns
select(Condition, LB_Good, LB_Poor, LB_Motivated, Age, Gender)
```

```r
# Convert Condition and Gender column to a factor (category)
d$Condition <- as.factor(d$Condition)
d$Gender <- as.factor(d$Gender)
```

# Descriptives

## Number of individuals in each group

```r
# Calculate number of participants in each condition
CountTable = d %>%
  select(Condition) %>%
  count(Condition) %>%
  mutate(Percentage = n / sum(n) * 100) %>%
  # Round the percentage to 2 decimal places
  mutate(Percentage = round(Percentage, 2)) %>%
  # Combine the counts and percentages into a single column called n (%) using
  unite("n (%)", n, Percentage, sep = " (") %>%
  mutate(`n (%)` = paste0(`n (%)`, "%)")) %>%
  gt() %>%
  tab_header(
    title = "Counts and Percentages of Participants in Each Condition",
  )

# Display the table
display_html(as_raw_html(CountTable))
```

## Distribution of response variable by group

```r
# Create a summary table
distribution_table = d %>%
    tbl_summary(
        by = Condition,
        include = c(LB_Good, LB_Poor, LB_Motivated)) %>%
    modify_header(label = "**Variable**") %>%
    add_overall() %>%
    as_gt() %>%
    tab_header(title = "Summary of the Distribution of Scores by Condition")

# Display the table
display_html(as_raw_html(distribution_table))
```

## Means and Standard Deviations of response variable by group

```r
# Table of means and standard deviations by condition for each variable
means_table = d %>%
    group_by(Condition) %>%
    summarise(
        LB_Good_Mean = mean(LB_Good, na.rm = TRUE),
        LB_Good_SD = sd(LB_Good, na.rm = TRUE),
        LB_Poor_Mean = mean(LB_Poor, na.rm = TRUE),
        LB_Poor_SD = sd(LB_Poor, na.rm = TRUE),
        LB_Motivated_Mean = mean(LB_Motivated, na.rm = TRUE),
        LB_Motivated_SD = sd(LB_Motivated, na.rm = TRUE)
    ) %>%
    gt() %>%
    tab_header(title = "Means and Standard Deviations by Condition")

# Display the table
display_html(as_raw_html(means_table))
```

```r
# Overall means and standard deviations
means_table_overall = d %>%
    summarise(
        LB_Good_Mean = mean(LB_Good, na.rm = TRUE),
        LB_Good_SD = sd(LB_Good, na.rm = TRUE),
        LB_Poor_Mean = mean(LB_Poor, na.rm = TRUE),
        LB_Poor_SD = sd(LB_Poor, na.rm = TRUE),
        LB_Motivated_Mean = mean(LB_Motivated, na.rm = TRUE),
        LB_Motivated_SD = sd(LB_Motivated, na.rm = TRUE)
    ) %>%
    gt() %>%
    tab_header(title = "Overall Means and Standard Deviations")

# Display the table
display_html(as_raw_html(means_table_overall))
```

# Correlation Matricies

## Gender Correlations

```r
# Remove rows that have 3 as the gender
d = d[d$Gender != 3,]

# Make Condition a factor
d$Condition <- as.factor(d$Condition)
# Make gender a factor
d$Gender <- as.factor(d$Gender)
```

### Positive Self Evaluation (LB_Good) Gender Differences

```r
pos_self_eval_lm1 = lm(LB_Good ~ Condition*Gender, data = d)
summary(pos_self_eval_lm1)
```

### Negative Self Evaluation (LB_Poor) Gender Differences

```r
neg_self_eval_lm1 = lm(LB_Poor ~ Condition*Gender, data = d)
summary(neg_self_eval_lm1)
```

### PA Motivation (LB_Motivated) Gender Differences

```r
pa_motivation_lm1 = lm(LB_Motivated ~ Condition*Gender, data = d)
summary(pa_motivation_lm1)
```

```r
table1 = apaTables::apa.reg.table(pos_self_eval_lm1, table.number = 1,filename = "reg table 1.doc")
table2 = apaTables::apa.reg.table(neg_self_eval_lm1, table.number = 2,filename = "reg table 2.doc")
table3 = apaTables::apa.reg.table(pa_motivation_lm1, table.number = 3, filename = "reg table 3.doc")
```

## Overall Correlations

```r
# Correlation matrix for the variables
all_cor = d %>%
    select(LB_Good, LB_Poor, LB_Motivated, Age) %>%
    cor(method = "pearson", use = "pairwise.complete.obs")

print(digits = 2, all_cor)
```

# Notes

Variables:

- Condition: Rank participant saw themselves in on the leaderboard (1-6)

- LB_Good: “I seem to perform very well compared to the others in the ranking list.” (1 (disagree strongly) - 5 (agree strongly))

- LB_Poor: “I seem to perform very poorly compared to the others in the ranking list.” (1 (disagree strongly) - 5 (agree strongly))

- LB_Motivated: “I feel motivated to improve my physical activity (steps or minutes of exercise).” (1 (disagree strongly) - 5 (agree strongly))

- Age: Age of participant

- Gender: Gender of participant

# Simple Mediation Models

## Model 1

```r
process(d, #data
        y = "LB_Motivated", # Dependent variable
        x = "Condition", # Independent variable
        m = "LB_Good", # Mediator
        cov = c("Age", "Gender"), # Covariates
        mcx = 1, # Indicates categorical independent variable
        total = 1, # Gives more information about the total effect
        model = 4, # PROCESS model
        progress = 1, # Shows progress
        seed = seed, # Bootstrap seed
        stand = 1, # Standardized coefficients
        effsize = 1 # Effect size
        )
```

## Model 2

```r
process(d,
        y = "LB_Motivated",
        x = "Condition",
        m = "LB_Poor", 
        cov = c("Age", "Gender"),
        mcx = 1,
        total = 1,
        model = 4,
        progress = 1,
        seed = seed,
        stand = 1,
        effsize = 1)
```

# Parallel Mediation Model

## Model 3

```r
process(d,
        y = "LB_Motivated",
        x = "Condition",
        m = c("LB_Good", "LB_Poor"),
        cov = c("Age", "Gender"),
        mcx = 1,
        total = 1,
        model = 4,
        progress = 1,
        effsize= 1,
        seed = seed,
        stand = 1)
```

# Notes

Variables:

- Condition: Rank participant saw themselves in on the leaderboard (1-6)

- LB_Good: “I seem to perform very well compared to the others in the ranking list.” (1 (disagree strongly) - 5 (agree strongly))

- LB_Poor: “I seem to perform very poorly compared to the others in the ranking list.” (1 (disagree strongly) - 5 (agree strongly))

- LB_Motivated: “I feel motivated to improve my physical activity (steps or minutes of exercise).” (1 (disagree strongly) - 5 (agree strongly))

- Age: Age of participant

- Gender: Gender of participant

# Setup

```r
# Libraries
library(tidyverse)
library(flexplot)

# Figure Setup
options(repr.plot.width=10, repr.plot.height=16)
```

```r
# Load data
d = read.csv("/Users/rajharsora/Downloads/csv files/DHS cleaned.csv") %>%
    # Include only needed columns
    select(Condition, LB_Good, LB_Poor, LB_Motivated, Age, Gender)
```

```r
# Remove rows that have 3 as the gender
d = d[d$Gender != 3,]

# Convert Gender and Condition column to a factor (category)
d$Condition <- as.factor(d$Condition)
d$Gender <- as.factor(d$Gender)
```

# ANOVA + Post-hoc tests

## Positive Self-Evaluation (LB_Good)

### LM Results

```r
# Compute ANOVA
LB_Good_ANOVA <- lm(LB_Good ~ Condition + Gender + Age, data = d)
print(summary(LB_Good_ANOVA))
```

### ANOVA Results

```r
print(aov(LB_Good_ANOVA))
```

### Tukey's HSD & Effect Sizes

```r
# Compute effect sizes, mean differences, etc.
print("Tukeys, Effect sizes, Mean Differences:")
estimates(LB_Good_ANOVA)

# Compute Tukey post-hoc test (to get the p-values)
print("Tukey post-hoc test results (repeated with p-values):")
LB_Good_Tukey = TukeyHSD(aov(LB_Good_ANOVA), "Condition")
print(LB_Good_Tukey)
```

## Negative Self-Evaluation (LB_Poor)

### LM Results

```r
# Compute ANOVA
LB_Poor_ANOVA <- lm(LB_Poor ~ Condition + Age + Gender, data = d)
summary(LB_Poor_ANOVA)
```

### ANOVA Results

```r
print(aov(LB_Poor_ANOVA))
```

### Tukey's HSD & Effect Sizes

```r
# Compute effect sizes, mean differences, etc.
print("Tukeys, Effect sizes, Mean Differences:")
estimates(LB_Poor_ANOVA)

print("Tukey post-hoc test results (repeated with p-values):")
# Post-hoc test for LB_Poor (to get the p-values)
LB_Poor_tukey <- TukeyHSD(aov(LB_Poor_ANOVA), "Condition")
LB_Poor_tukey
```
